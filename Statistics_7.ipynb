### Task 1: Conditional Probability

**Question:**

Given the following data from an e-commerce app: out of 500 users, 120 bought phone accessories, 80 bought a phone, and 50 bought both, calculate the probability that a user bought a phone given they already bought accessories (P(Phone|Accessories)).

**Answer:**

Given:

* Total users = 500
* Users who bought phone accessories = 120
* Users who bought both = 50

Using the conditional probability formula:

P(Phone | Accessories) = 50 / 120

= 0.4167

= 41.67%

**Final Answer:** The probability that a user bought a phone given that they already bought accessories is 41.67%.


### Task 2: Conditional Probability in Flipkart

**Question:**

Imagine you are analyzing Flipkart's user data: 30% of users buy headphones, and among those, 40% later buy a smartphone. Calculate the conditional probability that a randomly selected user buys a smartphone, given they have already bought headphones.

**Answer:**

Given:

* Probability of buying headphones = 30%
* Probability of buying a smartphone given headphones = 40%

Using conditional probability:

P(Smartphone | Headphones) = 40 / 100

= 0.40

= 40%

**Final Answer:** The probability that a user buys a smartphone given that they have already bought headphones is 40%.

### Task 4: Bayes' Theorem in Fraud Detection

**Question:**

A digital wallet app detects fraud based on spending patterns. If 2% of all transactions are fraudulent, and the app’s alert system correctly flags 90% of fraudulent transactions but also falsely flags 5% of normal transactions, use Bayes’ Theorem to calculate the probability that a transaction is actually fraudulent when flagged by the alert system.

**Answer:**

Given:

* Probability of fraud = 2% = 0.02
* Probability of normal transaction = 98% = 0.98
* Probability of alert given fraud = 90% = 0.90
* Probability of alert given normal transaction = 5% = 0.05

**Step 1: Calculate the total probability of an alert.**

P(Alert) = (0.90 × 0.02) + (0.05 × 0.98)

= 0.018 + 0.049

= 0.067

**Step 2: Apply Bayes' Theorem.**

P(Fraud | Alert) = (0.90 × 0.02) / 0.067

= 0.018 / 0.067

= 0.2687

= 26.87%

**Final Answer:** The probability that a transaction is actually fraudulent when flagged by the alert system is 26.87%.
